Step 3: Categorical Feature Encoding

Step 3.1 — Churn

```python
df["Churn"] = df["Churn"].map({
    "Yes": 1,
    "No": 0
})

print(df["Churn"].value_counts())
```

Step 3.2 — Churn datatype check

```python
print("Churn datatype:", df["Churn"].dtype)
```

Step 3.3 — Categorical columns

```python
categorical_columns = df.select_dtypes(
    include=["object"]
).columns

print("Categorical columns:")
print(list(categorical_columns))
```

Step 3.4 — customerID

```python
df_model = df.drop(columns=["customerID"])

print("Shape after removing customerID:", df_model.shape)
print("Remaining columns:", df_model.columns.tolist())
```

Step 3.5 — One-Hot Encoding

```python
categorical_columns = df_model.select_dtypes(
    include=["object"]
).columns

df_encoded = pd.get_dummies(
    df_model,
    columns=categorical_columns,
    drop_first=True
)

print("Original shape:", df_model.shape)
print("Encoded shape:", df_encoded.shape)
```

Step 3.6 — Boolean columns check

```python
print(df_encoded.dtypes.value_counts())
```

```python
bool_columns = df_encoded.select_dtypes(include=["bool"]).columns

df_encoded[bool_columns] = df_encoded[bool_columns].astype(int)

print("Boolean columns converted to integers.")
print(df_encoded.dtypes.value_counts())
```

Step 3.8: Encoding verify.

```python
print("Final dataset shape:", df_encoded.shape)

print("\nMissing values:")
print(df_encoded.isnull().sum().sum())

print("\nData types:")
print(df_encoded.dtypes.value_counts())

print("\nTarget column values:")
print(df_encoded["Churn"].value_counts())
```

Step 3.9 — Cleaned dataset

```python
import os

os.makedirs("../data/processed", exist_ok=True)

output_file = "../data/processed/telco_person3_week1_cleaned_encoded.csv"

df_encoded.to_csv(
    output_file,
    index=False
)

print("Cleaned and encoded dataset saved successfully!")
print("File:", output_file)
```

Baseline Analytics

Step 4.1 — Basic customer statistics

```python
print("Total Customers:", len(df))

print("Churned Customers:", (df["Churn"] == 1).sum())
print("Retained Customers:", (df["Churn"] == 0).sum())

print(
    "Churn Rate: {:.2f}%".format(
        df["Churn"].mean() * 100
    )
)
```

Step 4.2 — Basic numerical statistics

```python
print(df[[
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]].describe())
```

Step 4.3 — Customer segments based on tenure

```python
df["tenure_segment"] = pd.cut(
    df["tenure"],
    bins=[-1, 12, 24, 48, 72],
    labels=[
        "0-12 months",
        "13-24 months",
        "25-48 months",
        "49-72 months"
    ]
)

print(df["tenure_segment"].value_counts().sort_index())
```

Step 4.4 — Segment-wise churn rate

```python
segment_churn = (
    df.groupby("tenure_segment", observed=False)["Churn"]
      .mean()
      .mul(100)
      .round(2)
)

print("Churn Rate by Tenure Segment:")
print(segment_churn)
```

Step 4.5 — Contract-wise churn

```python
contract_churn = (
    df.groupby("Contract")["Churn"]
      .mean()
      .mul(100)
      .round(2)
      .sort_values(ascending=False)
)

print("Churn Rate by Contract:")
print(contract_churn)
```

Step 4.6 — Save baseline analytics

```python
import os

os.makedirs("../data/processed", exist_ok=True)

baseline_summary = pd.DataFrame({
    "Metric": [
        "Total Customers",
        "Churned Customers",
        "Retained Customers",
        "Churn Rate (%)",
        "Average Tenure",
        "Average Monthly Charges",
        "Average Total Charges"
    ],
    "Value": [
        len(df),
        int((df["Churn"] == 1).sum()),
        int((df["Churn"] == 0).sum()),
        round(df["Churn"].mean() * 100, 2),
        round(df["tenure"].mean(), 2),
        round(df["MonthlyCharges"].mean(), 2),
        round(df["TotalCharges"].mean(), 2)
    ]
})

baseline_summary.to_csv(
    "../data/processed/person3_week1_baseline_analytics.csv",
    index=False
)

print("Baseline analytics saved successfully!")
print(baseline_summary)
```
