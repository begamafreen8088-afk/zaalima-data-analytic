## 3. Contract, Tenure and Churn Analysis

This section analyzes the relationship between customer contract type,
customer tenure, and churn behavior.

```python
df["Contract"].value_counts()
```

```python
sns.countplot(data=df, x="Contract", hue="Churn")

plt.title("Customer Churn by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Number of Customers")
plt.xticks(rotation=20)
plt.show()
```

### Contract Type Analysis

This analysis compares customer churn across different contract types.
The visualization helps identify whether customers with different
contract durations have different churn patterns.

```python
df["tenure"].describe()
```

```python
sns.boxplot(data=df, x="Churn", y="tenure")

plt.title("Customer Tenure by Churn Status")
plt.xlabel("Churn")
plt.ylabel("Tenure (Months)")
plt.show()
```

```python
df["TenureGroup"] = pd.cut(
    df["tenure"],
    bins=[0, 12, 24, 48, 72],
    labels=["0-12", "13-24", "25-48", "49-72"],
    include_lowest=True
)
```

```python
df["TenureGroup"].value_counts().sort_index()
```

```python
sns.countplot(data=df, x="TenureGroup", hue="Churn")

plt.title("Customer Churn by Tenure Group")
plt.xlabel("Tenure Group")
plt.ylabel("Number of Customers")
plt.show()
```

### Key Findings

The contract and tenure analysis was performed to understand how
customer relationship duration and contract type are associated with
churn.

The visualizations provide comparisons of churn across contract types
and different customer tenure groups.

```python
df["Contract"].value_counts()
```

```python
pd.crosstab(
    df["Contract"],
    df["Churn"],
    normalize="index"
) * 100
```

```python
sns.countplot(data=df, x="Contract", hue="Churn")

plt.title("Customer Churn by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Number of Customers")
plt.xticks(rotation=20)
plt.show()
```

### Contract Type Analysis

This analysis examines customer churn across different contract types.
The churn percentages and visualization help identify whether
contract duration is associated with customer churn.

```python
df["tenure"].describe()
```

```python
sns.boxplot(data=df, x="Churn", y="tenure")

plt.title("Customer Tenure by Churn Status")
plt.xlabel("Churn")
plt.ylabel("Tenure (Months)")
plt.show()
```

### Tenure Analysis

This analysis compares customer tenure between customers who churned
and customers who remained with the company. The boxplot helps identify
differences in customer relationship duration between the two groups.

```python
df["TenureGroup"] = pd.cut(
    df["tenure"],
    bins=[0, 12, 24, 48, 72],
    labels=["0-12", "13-24", "25-48", "49-72"],
    include_lowest=True
)
```

```python
df["TenureGroup"].value_counts().sort_index()
```

```python
sns.countplot(data=df, x="TenureGroup", hue="Churn")

plt.title("Customer Churn by Tenure Group")
plt.xlabel("Tenure Group (Months)")
plt.ylabel("Number of Customers")
plt.show()
```

### Key Findings

The contract and tenure analysis provides insights into customer churn
behavior.

- Contract type shows differences in churn behavior across customers.
- Tenure distribution differs between customers who churned and those
  who remained.
- Customers were grouped into different tenure ranges to compare churn
  patterns across relationship durations.
- Contract duration and customer tenure can therefore be considered
  important variables for further churn analysis.
